```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
import numpy as np
import matplotlib.pyplot as plt
from torchvision import datasets, transforms
import os
import random
from collections import defaultdict
import logging
import wandb

# --- Logging and Seed
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

def set_all_seeds(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)
        torch.backends.cudnn.deterministic = True
        torch.backends.cudnn.benchmark = False
    os.environ["PYTHONHASHSEED"] = str(seed)

set_all_seeds(42)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# --- Model
class SimpleCNN(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()
        self.conv1 = nn.Conv2d(1, 32, 3)
        self.conv2 = nn.Conv2d(32, 64, 3)
        self.fc1 = nn.Linear(64 * 5 * 5, 128)
        self.fc2 = nn.Linear(128, num_classes)

    def forward(self, x):
        x = F.relu(self.conv1(x))         # (26x26)
        x = F.max_pool2d(x, 2)            # (13x13)
        x = F.relu(self.conv2(x))         # (11x11)
        x = F.max_pool2d(x, 2)            # (5x5)
        x = x.view(-1, 64 * 5 * 5)
        x = F.relu(self.fc1(x))
        return self.fc2(x)

# --- Data
transform = transforms.Compose([transforms.ToTensor()])
mnist_train = datasets.MNIST(root="./data", train=True, download=True, transform=transform)
mnist_test = datasets.MNIST(root="./data", train=False, download=True, transform=transform)

# --- Query strategies
def random_sampling(model, x_pool, n):
    return np.random.choice(len(x_pool), size=n, replace=False)

def uncertainty_sampling(model, x_pool, n, batch_size=128):
    model.eval()
    entropy_scores = []
    with torch.no_grad():
        for i in range(0, len(x_pool), batch_size):
            batch = x_pool[i:i+batch_size].to(device)
            outputs = model(batch)
            probs = F.softmax(outputs, dim=1)
            entropy = -torch.sum(probs * torch.log(probs + 1e-10), dim=1)
            entropy_scores.append(entropy)
    entropy_scores = torch.cat(entropy_scores)
    return torch.topk(entropy_scores, k=n).indices.cpu().numpy()

def bayesian_sampling(model, x_pool, n, dropout_passes=10, batch_size=128):
    model.train()
    num_samples = len(x_pool)
    num_classes = 10
    probs_sum = torch.zeros((num_samples, num_classes)).to(device)

    for _ in range(dropout_passes):
        current_probs = []
        for i in range(0, num_samples, batch_size):
            batch = x_pool[i:i+batch_size].to(device)
            with torch.no_grad():
                out = model(batch)
                prob = F.softmax(out, dim=1)
            current_probs.append(prob)
        current_probs = torch.cat(current_probs, dim=0)
        probs_sum += current_probs

    avg_probs = probs_sum / dropout_passes
    entropy = -torch.sum(avg_probs * torch.log(avg_probs + 1e-10), dim=1)
    return torch.topk(entropy, k=n).indices.cpu().numpy()


def hybrid_sampling(model, x_labeled, x_pool, n):
    model.eval()
    with torch.no_grad():
        probs = F.softmax(model(x_pool.to(device)), dim=1)
        entropy = -torch.sum(probs * torch.log(probs + 1e-10), dim=1)
    x_flat = x_pool.view(len(x_pool), -1)
    x_l_flat = x_labeled.view(len(x_labeled), -1)
    mean_labeled = x_l_flat.mean(dim=0)
    diversity = torch.norm(x_flat - mean_labeled, dim=1)
    score = 0.5 * entropy + 0.5 * diversity
    return torch.topk(score, k=n).indices.cpu().numpy()

strategies = {
    "Random": random_sampling,
    "Uncertainty": uncertainty_sampling,
    "Bayesian": bayesian_sampling,
}

# --- Utility
def get_class_accuracy(model, test_loader):
    correct = defaultdict(int)
    total = defaultdict(int)
    model.eval()
    with torch.no_grad():
        for x, y in test_loader:
            x, y = x.to(device), y.to(device)
            out = model(x)
            preds = torch.argmax(out, dim=1)
            for p, t in zip(preds, y):
                total[int(t)] += 1
                correct[int(t)] += int(p == t)
    return {cls: correct[cls] / total[cls] if total[cls] > 0 else 0.0 for cls in total}

def train(model, loader, optimizer, criterion):
    model.train()
    for x, y in loader:
        x, y = x.to(device), y.to(device)
        optimizer.zero_grad()
        loss = criterion(model(x), y)
        loss.backward()
        optimizer.step()

# --- Experiment Config
query_size = 200
max_iterations = 50
initial_class = 0
threshold = 0.99
final_results = {}

wandb.login(
    key="5509e4778375d7a603e946cae346f3a939027ed9"
)

for strategy_name, strategy_fn in strategies.items():
    run = wandb.init(project="active-learning-mnist", name=strategy_name, reinit=True)
    wandb.config.update({"query_size": query_size, "max_iterations": max_iterations, "threshold": threshold})

    logger.info(f"\n=== Running strategy: {strategy_name} ===")
    included_classes = [initial_class]
    labeled_indices = np.random.choice(np.where(np.array(mnist_train.targets) == initial_class)[0], size=100, replace=False).tolist()
    unlabeled_indices = list(set(range(len(mnist_train))) - set(labeled_indices))

    labeled_counts = {i: 0 for i in range(10)}
    for idx in labeled_indices:
        labeled_counts[mnist_train.targets[idx].item()] += 1

    sample_needed = {}
    acc_over_time = {i: [] for i in range(10)}
    model = SimpleCNN().to(device)
    optimizer = optim.Adam(model.parameters(), lr=1e-3)
    criterion = nn.CrossEntropyLoss()

    for cls in range(1, 10):
        logger.info(f"\n🔁 Adding class {cls}")
        included_classes.append(cls)

        for iteration in range(1, max_iterations + 1):
            logger.info(f"→ Iteration {iteration} with classes {included_classes}")

            labeled_subset = torch.utils.data.Subset(mnist_train, labeled_indices)
            unlabeled_subset = torch.utils.data.Subset(mnist_train, unlabeled_indices)

            labeled_loader = torch.utils.data.DataLoader(labeled_subset, batch_size=64, shuffle=True)
            test_loader = torch.utils.data.DataLoader(
                [(x, y) for x, y in mnist_test if y in included_classes],
                batch_size=256, shuffle=False
            )

            train(model, labeled_loader, optimizer, criterion)

            accs = get_class_accuracy(model, test_loader)
            for c, a in accs.items():
                acc_over_time[c].append(a)
            logger.info("Class Accuracies: " + ", ".join(f"{k}: {v:.3f}" for k, v in accs.items()))
            wandb.log({f"accuracy_class_{k}": v for k, v in accs.items()})

            if accs.get(cls, 0.0) >= threshold:
                logger.info(f"✅ Class {cls} reached {threshold*100:.0f}% accuracy with {labeled_counts[cls]} labeled samples.")
                sample_needed[(strategy_name, cls)] = labeled_counts[cls]
                break

            x_pool_all = [(x, y, i) for i, (x, y) in zip(unlabeled_indices, mnist_train) if y == cls]
            if not x_pool_all:
                logger.info("⚠️ No more samples from class {} available in pool.".format(cls))
                continue

            x_pool = torch.stack([x for x, y, _ in x_pool_all])
            dataset_indices = [i for x, y, i in x_pool_all]

            if strategy_name == "Hybrid":
                x_labeled = torch.stack([mnist_train[i][0] for i in labeled_indices])
                indices = strategy_fn(model, x_labeled, x_pool, query_size)
            else:
                indices = strategy_fn(model, x_pool, query_size)

            mapped_indices = [dataset_indices[i] for i in indices]
            for idx in mapped_indices:
                labeled_counts[mnist_train.targets[idx].item()] += 1

            labeled_indices.extend(mapped_indices)
            unlabeled_indices = list(set(unlabeled_indices) - set(mapped_indices))

    final_results[strategy_name] = sample_needed
    run.finish()

# --- Plots
plt.figure(figsize=(10, 6))
for strat, cls_dict in final_results.items():
    cls_ids = sorted(cls_dict.keys(), key=lambda x: x[1])
    x = [cls for (_, cls) in cls_ids]
    y = [cls_dict[(strat, cls)] for (_, cls) in cls_ids]
    plt.plot(x, y, marker='o', label=strat)

plt.xlabel("New Class ID")
plt.ylabel("Labeled Samples Needed to Reach 95% Accuracy")
plt.title("Comparison of Query Strategies on Class Detection")
plt.legend()
plt.grid()
plt.tight_layout()
plt.savefig("samples_needed_per_class.png")
plt.show()

# Additional plot: Accuracy over time for each class
for cls in range(10):
    if any(cls in acc_over_time and acc_over_time[cls] for acc_over_time in [acc_over_time]):
        plt.plot(acc_over_time[cls], label=f"Class {cls}")

plt.xlabel("Iteration")
plt.ylabel("Accuracy")
plt.title("Class Accuracy over Time")
plt.legend()
plt.grid()
plt.tight_layout()
plt.savefig("accuracy_over_time.png")
plt.show()
```
